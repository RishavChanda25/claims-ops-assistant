```python
import sys
import os

# Add the parent directory to Python's path so we can import 'src'
sys.path.append(os.path.abspath('..'))

from src.retrieval import retrieve_context
from src.generation import generate_answer

print("1. Starting End-to-End Multimodal Pipeline Test...\n")

test_query = "My car was parked outside and got caught in a massive flood. The car is ruined, and my $2,000 laptop was in the backseat and got destroyed by the water too. What is covered?"
print(f"User Query: '{test_query}'\n")

# Step 1: Retrieve text context based on the query string
print("2. Retrieving relevant clauses...")
docs = retrieve_context(test_query)
print(f"   -> Found {len(docs)} relevant chunks from the database.\n")

# Step 2: Load the local test image into memory as raw bytes
image_path = "test_damage.jpg"  # Ensure this file exists in your folder!
print(f"3. Loading test image: '{image_path}'...")

try:
    with open(image_path, "rb") as image_file:
        image_bytes = image_file.read()
    print("   -> Image loaded successfully into memory.\n")
except FileNotFoundError:
    print(f"❌ Error: Could not find '{image_path}'. Please place an image in this directory.")
    sys.exit(1)

# Step 3: Generate the multimodal claims report
print("4. Generating Multimodal Claims Assessment (Gemini 2.5 Flash)...")
final_answer = generate_answer(test_query, docs, image_bytes=image_bytes)

print("\n================ FINAL MULTIMODAL CLAIMS ASSESSMENT ================\n")
print(final_answer)
```
