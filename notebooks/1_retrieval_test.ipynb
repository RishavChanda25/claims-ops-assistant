```python
import sys
import os

# Add the parent directory to Python's path so we can import 'src'
sys.path.append(os.path.abspath('..'))

from src.retrieval import retrieve_context

print("1. Initializing Multi-Document Retriever...\n")

# The Ultimate Cross-Document Test Query
test_query = "My car was parked outside and got caught in a massive flood. The car is ruined, and my $2,000 laptop was in the backseat and got destroyed by the water too. What is covered?"

print(f"2. Querying Knowledge Base for: \n'{test_query}'\n")

# Call our encapsulated function
# By default, it uses k=5 and distance_threshold=1.1
try:
    retrieved_docs = retrieve_context(query=test_query)
    
    print(f"✅ Success! Retrieved {len(retrieved_docs)} highly relevant clauses.\n")
    print("--- TOP MATCHES RETRIEVED ---\n")
    
    for i, doc in enumerate(retrieved_docs):
        # Extract the filename so we know which policy we hit
        source_file = doc.metadata.get('source', 'Unknown').split('/')[-1].split('\\')[-1]
        
        print(f"Match {i+1} | Source: {source_file}")
        # Print the first 250 characters of the chunk to verify quality
        print(f"Snippet: {doc.page_content[:250].replace('  ', ' ')}...\n")
        print("-" * 60)
        
except Exception as e:
    print(f"❌ Error: {e}")
    print("Hint: Make sure you ran 0_ingestion_test.ipynb first to build the database!")
```
